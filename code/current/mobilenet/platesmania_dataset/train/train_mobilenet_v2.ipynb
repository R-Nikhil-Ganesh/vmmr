## 2. Configuration & Hyperparameters


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().resolve().parents[1]))  # mobilenet/ (paths.py)
import paths  # machine-specific paths: see paths.sh / paths.local.sh

IMG_SIZE = 512
BATCH_SIZE = 32
EPOCHS = 15
LR = 5e-4
BACKBONE_LR = 5e-5
DROPOUT = 0.25

CSV_PATH = paths.PM_MANIFEST_CSV
BASE_IMG_DIR = paths.PM_IMG_DIR
LABEL_MAP_PATH = paths.PM_LABEL_MAP
OUTPUT_DIR = Path("output_mobilenet_v2")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

## 3. Launch Training
Run the cell below, or use the headless runner (`run_train.sh` / `run_train_external.sh`):


```python
!python train_mobilenet_v2.py \
    --csv-path "{CSV_PATH}" \
    --base-img-dir "{BASE_IMG_DIR}" \
    --label-map-path "{LABEL_MAP_PATH}" \
    --output-dir "{OUTPUT_DIR}" \
    --img-size {IMG_SIZE} \
    --batch-size {BATCH_SIZE} \
    --epochs {EPOCHS} \
    --lr {LR} \
    --backbone-lr {BACKBONE_LR} \
    --dropout {DROPOUT} \
    --crop-top-pct 0.15 \
    --crop-bottom-pct 0.0
```
