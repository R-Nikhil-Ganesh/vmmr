# MobileNetV2 Training Pipeline: PlatesMania + External Merged (Model C)
---
Joint cross-domain training pipeline across 1,235 fine-grained Make/Model classes and 35 automotive makes, combining PlatesMania (top 15% banner crop) and External datasets (BoxCars116k, Stanford Cars, CompCars SV with bounding-box/bottom crop), domain-balanced sampling, GPU augmentation, and automated ONNX export.


## 1. Environment & GPU Verification


```python
import torch
import sys
from pathlib import Path

print(f"PyTorch: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
```

## 2. Configuration & Hyperparameters


```python
import sys
from pathlib import Path

DATASET_DIR = Path.cwd().resolve().parent
ROOT_MOBILENET_DIR = DATASET_DIR.parent
if str(ROOT_MOBILENET_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_MOBILENET_DIR))
import paths

IMG_SIZE = 512
BATCH_SIZE = 32
EVAL_BATCH_SIZE = 64
EPOCHS = 15
LR = 5e-4
BACKBONE_LR = 5e-5
DROPOUT = 0.25
EXT_FRACTION = 0.2

PM_CSV_PATH = paths.PM_MANIFEST_CSV
PM_IMG_DIR = paths.PM_IMG_DIR
LABEL_MAP_PATH = paths.PM_LABEL_MAP
EXT_SPLITS_DIR = DATASET_DIR / "splits_1235models"
OUTPUT_DIR = DATASET_DIR / "output_merged_run4"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

## 3. Launch Training
Run the cell below, or use the headless tmux runner (`bash bash_scripts/run_train_merged.sh`):


```python
!python train_mobilenet_v2_merged.py \
    --pm-csv-path "{PM_CSV_PATH}" \
    --pm-img-dir "{PM_IMG_DIR}" \
    --label-map-path "{LABEL_MAP_PATH}" \
    --ext-splits-dir "{EXT_SPLITS_DIR}" \
    --output-dir "{OUTPUT_DIR}" \
    --ext-fraction {EXT_FRACTION} \
    --img-size {IMG_SIZE} \
    --batch-size {BATCH_SIZE} \
    --eval-batch-size {EVAL_BATCH_SIZE} \
    --epochs {EPOCHS} \
    --lr {LR} \
    --backbone-lr {BACKBONE_LR} \
    --dropout {DROPOUT} \
    --aug-strength strong \
    --crop-jitter 0.5
```
